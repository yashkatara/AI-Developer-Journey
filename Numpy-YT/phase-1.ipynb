## Numpy array and basics


```python
import numpy as np
```

```python
arr_1d = np.array([1, 2, 3, 4, 5])
print(arr_1d)
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print(arr_2d)
```

#### list vs numpy array

```python
py_list = [1, 2, 3, 4, 5]
print ('Python list multiplication by 2:', py_list * 2)
np_array = np.array([[1,2,3]])
print ('Numpy array multiplication by 2:', np_array * 2)
```

##### Create array from scratch

```python
zeros = np.zeros((2, 3))
print('2x3 array of zeros:\n', zeros)

ones = np.ones((3, 2))
print('3x2 array of ones:\n', ones)

full = np.full((2, 2), 7)
print('2x2 array filled with 7:\n', full)

random = np.random.rand(2, 3)
print('2x3 array of random numbers:\n', random)
sequuence = np.arange(0, 10, 2)
print('Sequence from 0 to 10 with step 2:\n', sequuence)
```

#### Vector ,Matrix,tensor


```python
vector = np.array([1, 2, 3])
print('Vector:\n', vector)

matrix = np.array([[1, 2], [3, 4]])
print('Matrix:\n', matrix)

tensor = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print('Tensor:\n', tensor)
```

### Array Properties

```python
arr = np.array([[1,2,3],[4,5,6]])
print('Shape', arr.shape)
print('Number of dimensions', arr.ndim)
print('Size', arr.size)
print('Data type', arr.dtype)
```
